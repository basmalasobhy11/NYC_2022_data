```python
df['created_date'] = pd.to_datetime(df['created_date'])
```

```python
df['latitude'] = pd.to_numeric(df['latitude'], errors='coerce')
df['longitude'] = pd.to_numeric(df['longitude'], errors='coerce')
```

```python
df['latitude'].fillna(df['latitude'].median(), inplace=True)
df['longitude'].fillna(df['longitude'].median(), inplace=True)
```

```python
import time
from geopy.geocoders import Nominatim
from geopy.exc import GeocoderTimedOut, GeocoderUnavailable

# Initialize geocoder
geolocator = Nominatim(user_agent="zip_lookup")

# Safe reverse function with retry + delay
def safe_reverse(lat, lon, retries=3, delay=1):
    for attempt in range(retries):
        try:
            return geolocator.reverse((lat, lon), exactly_one=True, timeout=10)
        except (GeocoderTimedOut, GeocoderUnavailable):
            time.sleep(delay)
    return None

# Function to get ZIP from lat/lon
def get_zip(lat, lon):
    location = safe_reverse(lat, lon)
    if location and 'postcode' in location.raw['address']:
        return location.raw['address']['postcode']
    return None

# --- Step 1: Select only rows with NaN incident_zip ---
mask = df['incident_zip'].isna()
subset = df.loc[mask, ['latitude', 'longitude']].copy()

# --- Step 2: Apply geocoding only to NaNs ---
zip_codes = []
for i, row in subset.iterrows():
    zip_codes.append(get_zip(row['latitude'], row['longitude']))
    time.sleep(1)  # respect Nominatim rate limit

subset['incident_zip'] = zip_codes

# --- Step 3: Merge back into original DataFrame ---
df.loc[mask, 'incident_zip'] = subset['incident_zip']

print(df.head())
```

```python
import requests
import time

def get_bbl(lat, lon):
    url = f"https://api.nyc.gov/geoclient/v1/latlon.json?lat={lat}&lon={lon}&app_id=YOUR_APP_ID&app_key=YOUR_APP_KEY"
    response = requests.get(url).json()
    if 'latLon' in response and 'bbl' in response['latLon']:
        return response['latLon']['bbl']
    return None

mask = df['bbl'].isna()
subset = df.loc[mask, ['latitude', 'longitude']].copy()

bbl_values = []
for i, row in subset.iterrows():
    bbl_values.append(get_bbl(row['latitude'], row['longitude']))
    time.sleep(0.5) 

subset['bbl'] = bbl_values

df.loc[mask, 'bbl'] = subset['bbl']

print(df.head())



```
