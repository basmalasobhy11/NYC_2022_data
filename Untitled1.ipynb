```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib as plt
```

```python
df= pd.read_csv('nyc_311_data_2022.csv')
```

```python
df
```

```python
df.drop(['closed_date','city','borough','descriptor','agency_name','landmark'], axis=1, inplace=True)
```

```python
df['location_type'].fillna(df['location_type'].mode, inplace=True)
```

```python
df['created_date'] = pd.to_datetime(df['created_date'])
```

```python
top5 = df['complaint_type'].value_counts().head(5).index
subset = df[df['complaint_type'].isin(top5)]
pivot = subset.pivot_table(
    index='complaint_type',
    columns='location_type',
    
    values='unique_key',
    aggfunc='count',
    fill_value=0
)

plt.figure(figsize=(10, 9))
sns.heatmap(pivot, annot=True, fmt='d', cmap='Blues')
plt.title("Top 5 Complaint Types vs Location Type")
plt.show()
```
